## 1. Importing Libraries

Start by importing the necessary Python libraries:

```python
# Data manipulation
import numpy as np
import pandas as pd

# Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Statistics
from scipy import stats

# Configure plotting
plt.style.use('seaborn-v0_8')
sns.set_palette('husl')
%matplotlib inline
```

## 2. Loading Data

Load your data from the `data/raw/` directory:

```python
# Example: Loading a CSV file
# df = pd.read_csv('../data/raw/your_data.csv')

# Example: Creating sample data for demonstration
np.random.seed(42)
df = pd.DataFrame({
    'A': np.random.randn(100),
    'B': np.random.randn(100),
    'C': np.random.choice(['X', 'Y', 'Z'], 100)
})

print(f"Dataset shape: {df.shape}")
df.head()
```

## 3. Basic Data Exploration

Perform initial data exploration:

```python
# Dataset info
print("Dataset Information:")
df.info()
```

```python
# Summary statistics
print("\nSummary Statistics:")
df.describe()
```

```python
# Check for missing values
print("\nMissing Values:")
df.isnull().sum()
```

```python
# Basic visualization
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Histogram
df['A'].hist(bins=20, ax=axes[0])
axes[0].set_title('Distribution of Variable A')
axes[0].set_xlabel('Value')
axes[0].set_ylabel('Frequency')

# Scatter plot
axes[1].scatter(df['A'], df['B'], alpha=0.5)
axes[1].set_title('Relationship between A and B')
axes[1].set_xlabel('Variable A')
axes[1].set_ylabel('Variable B')

plt.tight_layout()
plt.show()
```
